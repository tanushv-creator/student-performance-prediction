# Student Performance Analysis
This notebook explores the student dataset, visualizes key patterns, and builds predictive models for final grade (G3) and pass/fail classification.

## 1. Imports & Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv("../data/processed/student_clean.csv")
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
df.head()
```

## 2. Data Overview

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing values:\n", df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
```

## 3. Exploratory Data Analysis (EDA)

### 3.1 Target Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["G3"], bins=20)
plt.title("Final Grade Distribution")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

### 3.2 Study Time vs Performance

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="studytime", y="G3", data=df)
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time (1=low, 4=high)")
plt.ylabel("Final Grade")
plt.tight_layout()
plt.show()
```

### 3.3 Correlation Heatmap

```python
plt.figure(figsize=(10, 7))
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### 3.4 Absences vs Performance

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="absences", y="G3", data=df)
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.tight_layout()
plt.show()
```

### 3.5 Pass / Fail Distribution

```python
df["result"] = df["G3"].apply(lambda x: "Pass" if x >= 10 else "Fail")

plt.figure(figsize=(6, 4))
sns.countplot(x="result", data=df)
plt.title("Pass vs Fail Count")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

## 4. Modelling

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error,
    accuracy_score, precision_score, recall_score, f1_score
)

X = df.drop(["G3", "result"], axis=1)
y_reg = df["G3"]
y_cls = df["result"]
```

### 4.1 Regression — Predicting Final Grade (G3)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)

reg_model = LinearRegression()
reg_model.fit(X_train, y_train)
y_pred_reg = reg_model.predict(X_test)

mae  = mean_absolute_error(y_test, y_pred_reg)
rmse = np.sqrt(mean_squared_error(y_test, y_pred_reg))

print("--- Regression Metrics ---")
print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
```

### 4.2 Classification — Predicting Pass / Fail

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y_cls, test_size=0.2, random_state=42
)

clf = RandomForestClassifier(random_state=42)
clf.fit(X_train, y_train)
y_pred_cls = clf.predict(X_test)

print("--- Classification Metrics ---")
print(f"Accuracy:  {accuracy_score(y_test, y_pred_cls):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_cls, average='weighted'):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred_cls, average='weighted'):.4f}")
print(f"F1 Score:  {f1_score(y_test, y_pred_cls, average='weighted'):.4f}")
```
