```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/student_clean.csv")
print(df.shape)
df.head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["G3"], bins=20, color="steelblue")
plt.title("Final Grade Distribution")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="studytime", y="G3", data=df, palette="Blues")
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time (1=low, 4=high)")
plt.ylabel("Final Grade")
plt.show()
```

```python
plt.figure(figsize=(10,7))
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="absences", y="G3", data=df, color="coral")
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()
```

```python
df["result"] = df["G3"].apply(lambda x: "Pass" if x >= 10 else "Fail")

plt.figure(figsize=(6,4))
sns.countplot(x="result", data=df, palette={"Pass": "steelblue", "Fail": "coral"})
plt.title("Pass vs Fail Count")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()
```
